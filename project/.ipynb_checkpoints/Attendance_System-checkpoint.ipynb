```python
import cv2
import os

def generate_dataset():
    # Load the cascade classifier
    face_classifier = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_frontalface_default.xml")
    
    if face_classifier.empty():
        print("Error: Could not load face cascade classifier.")
        return
    
    # Function to detect and crop faces
    def face_cropped(img):
        gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
        faces = face_classifier.detectMultiScale(gray, 1.3, 5)
        
        if len(faces) == 0:  # If no faces are found
            return None
        
        for (x, y, w, h) in faces:
            cropped_face = img[y:y+h, x:x+w]
            return cropped_face
    
    # Start capturing video from the webcam
    cap = cv2.VideoCapture(0)  # Using 0 to default to the first camera
    
    if not cap.isOpened():
        print("Error: Could not open webcam.")
        return
    
    user_id = input("Enter user ID (name): ")
    dataset_dir = "data"
    
    # Create the main data directory if it doesn't exist
    if not os.path.exists(dataset_dir):
        os.makedirs(dataset_dir)

    img_id = 0
    print("Press 'Enter' to capture images or 'Esc' to exit...")
    
    while True:
        ret, frame = cap.read()
        if not ret:
            print("Error: Could not read frame from webcam.")
            break
        
        face = face_cropped(frame)
        if face is not None:
            img_id += 1
            face_resized = cv2.resize(face, (200, 200))
            face_gray = cv2.cvtColor(face_resized, cv2.COLOR_BGR2GRAY)
            
            # Save the cropped face image directly in the 'data' directory
            file_name_path = os.path.join(dataset_dir, f"{user_id}_{img_id}.jpg")
            cv2.imwrite(file_name_path, face_gray)
            
            # Display the image with the image ID as an overlay
            cv2.putText(face_gray, str(img_id), (50, 50), cv2.FONT_HERSHEY_PLAIN, 1, (0, 255, 0), 2)
            cv2.imshow("Cropped Face", face_gray)

        # Check for key presses
        key = cv2.waitKey(1)
        if key == 13:  # Enter key
            print("Capturing stopped by user.")
            break
        elif key == 27:  # Esc key
            print("Exiting...")
            break

        # Automatically terminate after collecting 3000 samples
        if img_id >= 3000:
            print("Maximum samples collected (3000 images).")
            break

    # Release resources
    cap.release()
    cv2.destroyAllWindows()
    print("Collecting samples is completed....")

# Call the function
generate_dataset()
```

```python
# Helper function to create labels
def my_label(image_name):
    # Extract the user name from the file name (e.g., 'Nishchal_1.jpg' -> 'Nishchal')
    name = image_name.split('_')[0]  # Split the filename by underscore and get the first part (user ID)
    
    # Assign labels based on the name
    if name == "Nishchal":
        return np.array([1, 0, 0])
    elif name == "Sishir":
        return np.array([0, 1, 0])
    elif name == "Roshan":
        return np.array([0, 0, 1])
    else:
        return None  # In case of an unknown name

# Helper function to load and preprocess dataset
def my_data():
    data = []
    dataset_dir = "data"
    
    # Iterate through each image in the 'data' directory
    for img_file in tqdm(os.listdir(dataset_dir)):
        path = os.path.join(dataset_dir, img_file)
        
        # Read the image in grayscale
        img_data = cv2.imread(path, cv2.IMREAD_GRAYSCALE)
        
        # Make sure the image is valid and label exists
        label = my_label(img_file)
        if img_data is not None and label is not None:
            img_data = cv2.resize(img_data, (50, 50))  # Resize the image to 50x50
            data.append([np.array(img_data), label])  # Append the image and its label
    
    shuffle(data)  # Shuffle the data to ensure randomness
    return data

# Load the data
data = my_data()
```

```python
import cv2
import numpy as np
from tensorflow.keras.models import load_model
import os
import csv
import datetime

# Load the Haar Cascade face detector
face_classifier = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')

# Load the CNN face recognition model (already trained with data augmentation)
model = load_model('face_recognition_model_augmented.h5')

# Create a label-to-name mapping (ensure this matches the training labels)
label_map = {0: "Nishchal", 1: "Sishir", 2: "Roshan"}

# **Increase the confidence threshold**
threshold = 0.95  # Set a higher threshold (80%)

# Function to preprocess the face for CNN
def preprocess_face(face_img):
    face_resized = cv2.resize(face_img, (50, 50))  # Resize to match the input shape of the CNN
    face_resized = face_resized / 255.0  # Normalize pixel values
    face_reshaped = np.reshape(face_resized, (1, 50, 50, 1))  # Reshape for model input
    return face_reshaped

# Initialize attendance logging (use a CSV file for simplicity)
attendance_file = 'attendance_log.csv'
if not os.path.exists(attendance_file):
    with open(attendance_file, mode='w') as f:
        writer = csv.writer(f)
        writer.writerow(['Name', 'Date', 'Time'])  # Header row

# Function to log attendance
def log_attendance(name):
    current_time = datetime.datetime.now()
    date = current_time.date()
    time = current_time.time()

    # Append the attendance entry to the CSV file
    with open(attendance_file, mode='a') as f:
        writer = csv.writer(f)
        writer.writerow([name, date, time])

# Start video capture
cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()  # Capture frame from the webcam
    if not ret:
        break
    
    gray_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)  # Convert to grayscale
    faces = face_classifier.detectMultiScale(gray_frame, 1.3, 5)  # Detect faces
    
    for (x, y, w, h) in faces:
        face = gray_frame[y:y+h, x:x+w]  # Crop the face
        face_preprocessed = preprocess_face(face)  # Preprocess for CNN

        # Predict the person using the CNN model
        prediction = model.predict(face_preprocessed)
        confidence = np.max(prediction)  # Get the highest confidence/probability
        predicted_index = np.argmax(prediction)  # Get the predicted label index
        
        # **Only recognize face if confidence is higher than the threshold**
        if confidence > threshold:
            predicted_name = label_map[predicted_index]  # Map it to the person's name
            log_attendance(predicted_name)  # Log the attendance
        else:
            predicted_name = "Unknown"  # Mark as unknown if below the threshold

        # Draw a rectangle around the face and display the predicted name
        cv2.rectangle(frame, (x, y), (x + w, y + h), (255, 0, 0), 2)
        cv2.putText(frame, predicted_name, (x, y - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.9, (255, 255, 0), 2)
    
    # Display the frame with the face detection and recognition
    cv2.imshow('Real-Time Face Recognition', frame)
    
    # Break the loop on 'q' key press
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Release the video capture and close windows
cap.release()
cv2.destroyAllWindows()
```
